```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df = pd.read_csv("gen_data.csv")

print(df.head(10))
```

```python
n=df.isnull()
ns=n.sum()
ds=df.shape[0]
percent_missing = (ns / ds) * 100
print(percent_missing)
print(percent_missing.sum())
#if the missing values are above 50% then we can drop the column
```

```python
print("Shape:", df.shape)
print("__________________________missingvalues________________________________________")
print(df.isnull().sum())
print("__________________________data type  ________________________________________")
print(df.dtypes)
print("__________________________duplicate rows________________________________________")
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("after removing duplicates", df.shape)
```

```python

num_cols = df.select_dtypes(include="number").columns
categ_cols = df.select_dtypes(include=["str", "object", "category"]).columns

'''value handling using pd'''
for column in num_cols:
    df[column] = df[column].fillna(df[column].median())

for column in categ_cols:
    df[column] = df[column].fillna(df[column].mode()[0])

print("_______________________Missing values after cleaning__________________________")
print(df.isnull().sum())
```

```python
num_cols = df.select_dtypes(include="float64").columns
scaler = StandardScaler()

df[num_cols] = scaler.fit_transform(df[num_cols])

print(df[num_cols].head(10))
```

```python
df = pd.get_dummies(df, columns=categ_cols, dtype=int)
print(df.head(10))
```

```python
print("NEW row and cols", df.shape)
print("---------------------------Final Missing Value--------------------")
print(df.isnull().sum())

df.to_csv("cleandt.csv", index=False)

print("Cleaned dataset saved successfully!")
```
